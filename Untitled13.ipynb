```python
engine = pyttsx3.init()

# List available voices
voices = engine.getProperty('voices')
for voice in voices:
    print(f"ID: {voice.id}, Name: {voice.name}, Gender: {voice.gender}")

# Set the engine to use a male voice
engine.setProperty('voice', voices[0].id)  # Adjust index based on available voices

# Set speech rate
engine.setProperty('rate', 150)

# Set volume
engine.setProperty('volume', 1.0)

# Speak the text
engine.say("Hello, this is a test of the text-to-speech function.")
engine.runAndWait()
```
