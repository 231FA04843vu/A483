```python
import pandas as pd
import seaborn as sns
```

```python
dt=pd.read_csv('/content/Mall_Customers.csv')
```

```python
dt.info()
```

```python
for i in dt.columns:
  if dt[i].dtype=="object":
    dt[i]=dt[i].astype("category").cat.codes
```

```python
dt.corr()
```

```python
x=dt.iloc[:,3:]
x.head()
```

```python
from sklearn.cluster import KMeans
```

```python
from sklearn.cluster import KMeans
wss = []
for i in range(1,11):
  km = KMeans(n_clusters=i, init="k-means++", random_state=42, n_init=42)
  km.fit(x)
  wss.append(km.inertia_)
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1,11), wss)
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
kmeans.fit(x)
```

```python
labels=kmeans.labels_
dt["cluster"] =labels
```

```python
plt.scatter(dt["Annual Income (k$)"], dt["Spending Score (1-100)"], c=labels)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("mall customer segmentation")
plt.show()
```
